```python
# 행인덱스의 재배열

import pandas as pd

dict_data = {
    "c0" : [1, 2, 3],
    "c1" : [4, 5, 6],
    "c2" : [7, 8, 9],
    "c3" : [10, 11, 12],
    "c4" : [13, 14, 15]
}

df = pd.DataFrame(dict_data,index = ["r0","r1","r2"])
```

```python
df
```

```python
new_index = ["r0","r1","r2","r3","r4"]
ndf = df.reindex(new_index)
ndf
```

```python
ndf2 = df.reindex(new_index,fill_value=0)
ndf2
```

```python
# [문제 1] Series 생성 및 변수명 지정 

# 다음 데이터를 활용하여 Pandas Series를 생성하세요. 
# 생성할 때 데이터의 의미가 잘 드러나도록 명확한 변수명을 지정해야 합니다.

# [85, 92, 78, 90, 88] 리스트를 사용하여 5명의 학생 점수를 담은 Series를 만드세요.

# 위에서 만든 Series에 ['Alice', 'Bob', 'Charlie', 'David', 'Eve']라는 문자열 리스트를 인덱스(Index)로 지정하세요.

# 과목별 과제 점수를 의미하는 딕셔너리 {'Python': 95, 'SQL': 85, 'Web': 90}를 활용하여 또 다른 Series를 생성하세요.


import pandas as pd

lis = [85, 92, 78, 90, 88]

sr = pd.Series(lis)
print(sr)
print()

sr1 = pd.Series(lis,index = ['Alice', 'Bob', 'Charlie', 'David', 'Eve'] )
print(sr1)
print()

dic_data = {'Python': 95, 'SQL': 85, 'Web': 90}

sr2 = pd.Series(dic_data)

print(sr2)
```
