```python
## isin() 매서드 활ㅇㅇ
## DF객체.isin(추출 값의 리스트)
import pandas as pd
import seaborn as sns

titanic = sns.load_dataset("titanic")

pd.set_option("display.max_columns",10)

titanic.head()
```

```python
mask1 = titanic["embark_town"] == "Southhampton"
mask2 = titanic["embark_town"] == "Queenstown"

df_boolean = titanic[mask1|mask2]
df_boolean
```

```python
isin_filter = titanic["embark_town"].isin(["Southhampton","Queenstown"])

df_isin = titanic[isin_filter]
df_isin
```
