```python
#시리즈

## 외부라이브러리 pandas 설치
## pip install pandas


## 라이브러리 불러오기
import pandas as pd


dict_data = {
    "a" : 1,
    "b" : 2,
    "c" : 3,

}

sr = pd.Series(dict_data)

print(type(sr))
```

```python
print(sr) 
```

```python
## 인덱스 배열
## Series객체.index 

print(sr.index)
```

```python
## 데이터 값 배열
## Serieso객체.values

print(sr.values)
```

```python
## 시리즈 인덱스

import pandas as pd

list_data = ["2019-01-01", 3.14, "ABC", 100, True]

sr = pd.Series(list_data)

print(sr)



```

```python
## 인덱스

## values
 
print(sr.index)
print(sr.values)
```

```python
## 시리즈 객체의 데이터 타입 확인

print(sr.dtype)
```

```python
## 원소의 개수 확인하기

print(len(sr))
```

```python
## 배열의 형태

sr.shape
```

```python
## 데이터의 구조의 차원수
## ndim 속성
print(sr.ndim)
```
