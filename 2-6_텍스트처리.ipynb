```python
## 문자열 매서드

import pandas as pd
import numpy as np


list_data = [
    "Apple_사과",
    "Banana_바나나",
    "Cherry_체리",
    np.nan
]

ser = pd.Series(
    list_data, 
    index=["First "," Second"," Third", "Fourth" ],
    dtype="string"
    )
ser
```

```python
## 소문자로 변형 -> str접근자.lower()

print(ser.str.lower())
ser

## 대문자로 변형 -> str접근자.upper()


print(ser.str.upper())


## 문자열 길이 계산 -> str접근자.len()
# len(ser) -> 내장함수 len() :객체의 행의수를 구할때


print(ser.str.len())
```

```python
## 문자열 분할

exam1 = ser.str.split("_")
print(exam1)
type(exam1)
```

```python
## 문자열으로 분할하여 데이터프레임으로 화장
exam2 = ser.str.split("_",expand=True)
print(exam2)
print(type(exam2))
```

```python
## 문자열 중 특정 요소 선택

ser.str.split("_").str.get(1)
```

```python
## 행 인덱스 추출

idx = ser.index
print(idx)
```

```python
print(idx.str.strip())
```

```python
# 특정문자 포함여부 확인

print(ser)
print()
print(ser.str.contains("A"))
contains_A = ser.str.contains("A", na = False)
print(contains_A)

contains_B = ser.str.contains(r"[A|B][a-z]+",)
print(contains_B)
```
